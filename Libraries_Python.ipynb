```python
import pandas as pd
import numpy as np
```

**Question 1: Load the given retail sales dataset into a PandasData Frame and display hte first five rows along with the dataset information to understand its structure.**

**Answer:**

```python
import pandas as pd
df = pd.read_csv('/content/retail_dataset .csv')
print(df.head())
print(df.info())
```

**Question 2: Inspect the dataset and identify missing values in each column. Display the total number of missing values present in the dataset.**

**Answer:**

```python

print("Missing values per column:")
print(df.isnull().sum())


total_missing = df.isnull().sum().sum()
print(f"\nTotal missing values in dataset: {total_missing}")
```

**Question 3: Replace missing values in the Quantity and Price Columns with their respective mean values using Pandas.**

**Answer:**

```python

df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())
df['Price'] = df['Price'].fillna(df['Price'].mean())

print("Missing values after imputation:")
print(df[['Quantity', 'Price']].isnull().sum())
```

**Question 4: Remove all rows where the Product Category or Region Column contains missing values.**

**Answer:**

```python
# Remove rows with missing Product Category or Region
df = df.dropna(subset=['Product Category', 'Region'])

print("Shape after removing rows:", df.shape)
print("Missing values in Product Category and Region:")
print(df[['Product Category', 'Region']].isnull().sum())
```

**Question 5:** **Create a new column called Revenue using NumPy, where**
**Revenue** **=** **Quantity** * **Price**  
**Display the updated dataset with the new column.**

**Answer:**

```python
# Create Revenue column
df['Revenue'] = df['Quantity'] * df['Price']

print("Dataset with Revenue column:")
print(df.head())
```

**Question 6: Using NumPy, calculate the total revenue generated from all transactions in the dataset.**

**Answer:**

```python
# Total revenue using NumPy
total_revenue = np.sum(df['Revenue'])
print(f"Total Revenue: ${total_revenue:,.2f}")
```

**Question 7: Group the dataset by Product Category and calculate the tota revenue generated by each category.**

**Answer:**

```python
# Revenue by Product Category
category_revenue = df.groupby('Product Category')['Revenue'].sum().sort_values(ascending=False)
print("Revenue by Product Category:")
print(category_revenue)
```

**Question 8: Based on the revenue calculated for each product category, identify the top 3 categories with the highest revenue and the bottom 3 categories with the lowest revenue.**

**Answer:**

```python
# Top 3 and Bottom 3 categories
top_3 = category_revenue.head(3)
bottom_3 = category_revenue.tail(3)

print("Top 3 Categories by Revenue:")
print(top_3)
print("\nBottom 3 Categories by Revenue:")
print(bottom_3)
```

**Question 9: Group the dataset by Region and calculate the total revenue generated in each region. Identify the region with the highest revenue and the region with the lowest revenue.**

**Answer:**

```python
# Revenue by Region
region_revenue = df.groupby('Region')['Revenue'].sum().sort_values(ascending=False)

print("Revenue by Region:")
print(region_revenue)

print(f"\nHighest Revenue Region: {region_revenue.idxmax()} (${region_revenue.max():,.2f})")
print(f"Lowest Revenue Region: {region_revenue.idxmin()} (${region_revenue.min():,.2f})")
```

**Question 10: Extract the month from the Date Column using Pandas and group the dataset by month to calculate the total monthly revenue.Also, calculate the mean, median, and standard deviation of the Revenue Column using NumPy.**

**Answer:**

```python
# Extract month
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month

# Monthly revenue
monthly_revenue = df.groupby('Month')['Revenue'].sum()
print("Monthly Revenue:")
print(monthly_revenue)

# Statistics using NumPy
print(f"\nRevenue Statistics:")
print(f"Mean Revenue: ${np.mean(df['Revenue']):,.2f}")
print(f"Median Revenue: ${np.median(df['Revenue']):,.2f}")
print(f"Standard Deviation: ${np.std(df['Revenue']):,.2f}")
```
